#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
# split data into training and testing sets
march_split <- initial_split(tidy_march, prop = 0.75, strata = tourney_finish)
march_train <- training(march_split)
march_test  <- testing(march_split)

# standardize predictors for kNN
march_recipe <- recipe(tourney_finish ~ seed + reg_wins_pct, data = march_train) |>
step_scale(all_predictors()) |>
step_center(all_predictors())

#create kNN classification model
knn_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = 5) |>
set_engine("kknn") |>
set_mode("classification")

# fit model on training data
march_workflow <- workflow() |>
add_recipe(march_recipe) |>
add_model(knn_spec)
march_fit <- fit(march_workflow, data = march_train)

# predict on test data
march_predictions <- predict(march_fit, march_test) |>
bind_cols(march_test)

# evaluate model performance using accuracy
march_accuracy <- accuracy(march_predictions, truth = tourney_finish, estimate = .pred_class)
march_accuracy
```
